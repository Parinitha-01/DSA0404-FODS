```python

import pandas as pd
import re
from collections import Counter
import matplotlib.pyplot as plt

# Load the dataset from Downloads folder
df = pd.read_csv(r"/content/16bbc-text.csv")

# Extract text from the dataset
text_data = " ".join(df["text"].dropna().astype(str))

# Convert text into lowercase
text_data = text_data.lower()

# Extract words
words = re.findall(r'\b[a-z]+\b', text_data)

# Calculate word frequency
word_frequency = Counter(words)

# Display the word frequency
print("Word Frequency Distribution:\n")

for word, frequency in word_frequency.most_common(20):
    print(word, ":", frequency)

# Display the top 10 words in a bar chart
top_words = word_frequency.most_common(10)

words_list = [item[0] for item in top_words]
frequency_list = [item[1] for item in top_words]

plt.bar(words_list, frequency_list)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Word Frequency Distribution")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import re
import nltk
import matplotlib.pyplot as plt
from collections import Counter
from nltk.corpus import stopwords

# Download stopwords
nltk.download('stopwords')

# Load the dataset
df = pd.read_csv(r"/content/17customer_reviews.csv", encoding='latin1')

# Get English stop words
stop_words = set(stopwords.words('english'))

# Combine all customer reviews
text_data = " ".join(df["Review Text"].dropna().astype(str))

# Convert text into lowercase
text_data = text_data.lower()

# Remove punctuation and extract words
words = re.findall(r'\b[a-z]+\b', text_data)

# Remove stop words
filtered_words = [word for word in words if word not in stop_words]

# Calculate word frequency
word_frequency = Counter(filtered_words)

# Get N from user
N = int(input("Enter the number of top words to display: "))

# Display top N frequent words
top_words = word_frequency.most_common(N)

print("\nTop", N, "Most Frequent Words:\n")

for word, frequency in top_words:
    print(word, ":", frequency)

# Plot bar graph
words_list = [item[0] for item in top_words]
frequency_list = [item[1] for item in top_words]

plt.bar(words_list, frequency_list)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Top N Word Frequency Distribution")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
